## 参考实现 `w01`（含依赖导入）

```
import json
import random
import re
import sys
import numpy as np  # 通用科学计算


def w01():
    device_id = "DEV-2026-001"
    name = "智能注塑机"
    running = True
    temp = 56.5
    press_kpa = 450.0
    rpm = 1480

    def c_to_f(c):
        return c * 9 / 5 + 32

    print("====== 设备台账卡 ======")
    print(f"编号: {device_id}")
    print(f"名称: {name:<10} 状态: {'运行' if running else '停止'}")
    print(f"温度: {temp:>6.1f}℃ ({c_to_f(temp):.1f}℉)")
    print(f"压力: {press_kpa / 1000:.2f}MPa  转速: {rpm}rpm")
```
