## 参考实现 `w06`（含依赖导入）

```
import json
import random
import re
import sys
import numpy as np  # 通用科学计算


def w06():
    import logging
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s %(levelname)s %(message)s",
        handlers=[logging.StreamHandler(),
                  logging.FileHandler("w06.log", encoding="utf-8")])
    log = logging.getLogger("collector")

    raw = "temp,press\n56.2,0.45\nerr,0.46\n57.5,\n61.0,0.47\n58.3,0.44\n"
    ok = skip = fail = 0
    for i, line in enumerate(raw.splitlines()):
        if line.startswith("temp"):
            continue
        try:
            t, p = line.split(",")
            float(t), float(p)
            log.info("第%d条采集成功: %s", i, line)
            ok += 1
        except ValueError:
            log.warning("脏数据跳过: %r", line)
            skip += 1
        finally:
            pass
    print(f"采集统计: 成功{ok} 跳过{skip}")
```
